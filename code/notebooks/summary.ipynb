```python
import pandas as pd
import glob
import os
import numpy as np
```

```python
DATA_DIR = '../../results/'
```

```python
def load_df(files):
    df = pd.DataFrame()
    for fn in files:
        tmp = pd.read_csv(fn, index_col=None)
        df = pd.concat([df, tmp])
    df.sort_values(['chunk_id','main_paper_id'], inplace=True)
    df.rename(columns={'main_paper_id':'chunk_paper_id'}, inplace=True)
    return df
```

```python
df_publications = load_df(glob.glob(os.path.join(DATA_DIR,'chunk*publications.csv')))
df_references = load_df(glob.glob(os.path.join(DATA_DIR,'chunk*references.csv')))
df_citations = load_df(glob.glob(os.path.join(DATA_DIR,'chunk*citations.csv')))
```

```python
df_publications.shape, df_references.shape, df_citations.shape
```

```python
df_publications.loc[:, 'paper_id'] = np.arange(1,df_publications.shape[0]+1)
df_publications = df_publications[['paper_id', 'title', 'author', 'publication_date', 'source', 'database',
                                   'dbname', 'scode', 'sdb',
                                   'cites', 'downloads', 'result_rank', 'references_count', 'citations_count',
                                   'chunk_id', 'chunk_paper_id']]
df_publications.references_count = df_publications.references_count.astype(int)
df_publications
```

```python
df_references = df_references.merge(df_publications[['chunk_id','chunk_paper_id','paper_id']], left_on=['chunk_id','chunk_paper_id'], right_on=['chunk_id','chunk_paper_id'], left_index=False, right_index=False).rename(columns={'order':'reference_id'})
df_references = df_references[['paper_id', 'reference_id', 'title', 'authors', 'journal', 'year', 'kind',
                               'filename', 'dbcode', 'dbname', 'chunk_id', 'chunk_paper_id']]
df_references
```

```python
df_citations = df_citations.merge(df_publications[['chunk_id','chunk_paper_id','paper_id']], left_on=['chunk_id','chunk_paper_id'], right_on=['chunk_id','chunk_paper_id'], left_index=False, right_index=False).rename(columns={'order':'citation_id'})
df_citations = df_citations[['paper_id', 'citation_id', 'title', 'authors', 'journal', 'year', 'kind',
                             'filename', 'dbcode', 'dbname', 'chunk_id', 'chunk_paper_id']]
df_citations
```

```python
fn_pub = os.path.join(DATA_DIR, 'publications.csv')
df_publications.set_index('paper_id').to_csv(fn_pub, index=True)
fn_pub
```

```python
fn_ref = os.path.join(DATA_DIR, 'references.csv')
df_references.set_index(['paper_id','reference_id']).to_csv(fn_ref, index=True)
fn_ref
```

```python
fn_cit = os.path.join(DATA_DIR, 'citations.csv')
df_citations.set_index(['paper_id','citation_id']).to_csv(fn_cit, index=True)
fn_cit
```
